```python
import pandas as pd

df_zhvf_growth = pd.read_csv('../data/zhvf_growth.csv')
```

```python
df_zhvf_growth.head()
```

```python
dallas_zips = df_zhvf_growth[(df_zhvf_growth['City'] == "Dallas") & (df_zhvf_growth['State'] == "TX")]
```

```python
dallas_zips['RegionName']
```

```python
# --- Aggregate and sort counties by projected growth ---
# Assuming the columns represent projected growth values for different dates
growth_columns = ["2025-11-30", "2026-01-31", "2026-10-31"]  # Replace with actual column names

# Calculate the aggregated growth (e.g., sum or average) for each county
dallas_zips["Aggregated_Growth"] = dallas_zips[growth_columns].mean(axis=1)

# Sort counties by aggregated growth
sorted_counties = dallas_zips.sort_values(by="Aggregated_Growth", ascending=False)

# Display the top counties with the highest aggregated growth
print("Top counties with good aggregated projected growth:")
print(sorted_counties[["RegionName", "Aggregated_Growth"]].head(10).to_markdown(index=False))
```

```python
# --- Sort counties by bottom-tier ZHVI ---
# Assuming the columns represent bottom-tier ZHVI values for different dates
zhvi_columns = ["2025-11-30", "2026-01-31", "2026-10-31"]  # Replace with actual column names

# Calculate the average bottom-tier ZHVI for each county
dallas_zips["Average_Bottom_Tier_ZHVI"] = dallas_zips[zhvi_columns].mean(axis=1)

# Sort counties by the lowest average bottom-tier ZHVI
sorted_counties = dallas_zips.sort_values(by="Average_Bottom_Tier_ZHVI", ascending=True)

# Display the top counties with the lowest bottom-tier ZHVI
print("Top counties with the lowest bottom-tier ZHVI (affordable for first-time buyers):")
print(sorted_counties[["RegionName", "Average_Bottom_Tier_ZHVI"]].head(10).to_markdown(index=False))
```

```python
# --- Sort counties by mid-tier ZHVI ---
# Assuming the columns represent mid-tier ZHVI values for different dates
zhvi_columns = ["2025-11-30", "2026-01-31", "2026-10-31"]  # Replace with actual column names

# Calculate the average mid-tier ZHVI for each county
dallas_zips["Average_Mid_Tier_ZHVI"] = dallas_zips[zhvi_columns].mean(axis=1)

# Sort counties by the lowest average mid-tier ZHVI
sorted_counties = dallas_zips.sort_values(by="Average_Mid_Tier_ZHVI", ascending=True)

# Display the top counties with the lowest mid-tier ZHVI
print("Top counties with the lowest mid-tier ZHVI (affordable for mid-range buyers):")
print(sorted_counties[["RegionName", "Average_Mid_Tier_ZHVI"]].head(10).to_markdown(index=False))
```

```python
# --- Filter and sort counties by positive bottom-tier ZHVI aggregates ---
# Assuming the columns represent bottom-tier ZHVI values for different dates
zhvi_columns = ["2025-11-30", "2026-01-31", "2026-10-31"]  # Replace with actual column names

# Calculate the average bottom-tier ZHVI for each county
dallas_zips["Average_Bottom_Tier_ZHVI"] = dallas_zips[zhvi_columns].mean(axis=1)

# Filter counties with positive aggregates
dallas_zips_positive = dallas_zips[dallas_zips["Average_Bottom_Tier_ZHVI"] > 0]

# Sort counties by the lowest average bottom-tier ZHVI
sorted_counties = dallas_zips_positive.sort_values(by="Average_Bottom_Tier_ZHVI", ascending=False)

# Display the top counties with positive bottom-tier ZHVI aggregates
print("Top counties with positive bottom-tier ZHVI aggregates (affordable for first-time buyers):")
print(sorted_counties[["RegionName", "Average_Bottom_Tier_ZHVI"]].head(10).to_markdown(index=False))
```

```python
growth_cols = [c for c in df_zhvf_growth.columns if c.startswith("2025") or c.startswith("2026")]
```

```python
df_zhvf_growth["TotalGrowth"] = df_zhvf_growth[growth_cols].sum(axis=1)
```

```python
df_zhvf_growth["Stability"] = 1 / (df_zhvf_growth[growth_cols].std(axis=1) + 1e-6)  # Adding a small constant to avoid division by zero
```

```python

df_zhvf_growth["AvgGrowth"] = df_zhvf_growth[growth_cols].mean(axis=1)
```

```python
def normalize(x):
    return (x - x.min()) / (x.max() - x.min())

df_zhvf_growth["Growth_norm"] = normalize(df_zhvf_growth["AvgGrowth"])
df_zhvf_growth["Total_norm"]  = normalize(df_zhvf_growth["TotalGrowth"])
df_zhvf_growth["Stable_norm"] = normalize(df_zhvf_growth["Stability"])
```

```python
df_zhvf_growth["ZipScore"] = (
    0.50 * df_zhvf_growth["Growth_norm"] +
    0.30 * df_zhvf_growth["Stable_norm"] +
    0.20 * df_zhvf_growth["Total_norm"]
)
```

```python

df_dallas = df_zhvf_growth[(df_zhvf_growth['City'] == "Dallas") & (df_zhvf_growth['State'] == "TX")]
df_ranked = df_dallas.sort_values("ZipScore", ascending=False)

df_ranked[["RegionName", "CountyName", "ZipScore", "Growth_norm", "Stable_norm","Total_norm"]].head(20)
```
